# Horizonte lunar de Site01 (R2)

Prueba de concepto de la sección 9.3 con el DEM Site01 de 5 m y el COG polar de 80 m de [NASA/PGDA](https://pgda.gsfc.nasa.gov/products/78). La coordenada del catálogo es aproximada; no representa una coordenada de aterrizaje verificada. Ejecutar `data/scripts/prepare_dem.py` antes de abrir el notebook.


```python
import numpy as np
import matplotlib.pyplot as plt
from lumen.v1.common_pb2 import Site
from lumen_engine.dem.coverage import coverage, terrain_point
from lumen_engine.horizon.mask import compute

site = Site(lat=-89.45, lon=-137.31, obs_height_m=2)
print(coverage(site.lat, site.lon))
print('Terreno y pendiente:', terrain_point(site.lat, site.lon))
mask = compute(site, az_step_deg=0.25, max_range_km=200)
print('Capas usadas:', list(mask.layers_used))
print('Mín/máx elevación:', min(mask.elevation_deg), max(mask.elevation_deg))
```

```python
az = np.deg2rad(np.arange(len(mask.elevation_deg)) * mask.az_step_deg)
fig, ax = plt.subplots(figsize=(9, 5), subplot_kw={'projection': 'polar'})
ax.plot(az, mask.elevation_deg, color='#9b6b20', linewidth=1.1)
ax.set_theta_zero_location('N')
ax.set_theta_direction(-1)
ax.set_title('Site01 · horizonte lunar con curvatura · observador a 2 m')
ax.set_ylabel('Elevación del terreno [grados]')
plt.show()
```

Los tests `tests/unit/test_dem_horizon.py` verifican el caso plano contra el dip analítico (tolerancia 0.0001°) y una montaña de altura conocida. La convergencia radial está documentada en `docs/validation.md`; la comparación visual con QuickMap/Moon Trek sigue pendiente.
